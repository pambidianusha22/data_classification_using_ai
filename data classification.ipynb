# Import Libraries

```python
import pandas as pd

from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    f1_score
)
```

# Load Dataset

```python
# Load iris dataset

iris = load_iris()

# Convert to dataframe

data = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add target column

data['target'] = iris.target

# Display first 5 rows

data.head()
```

# Dataset Information

```python
print(data.info())

print(data.shape)

print(data.isnull().sum())
```

# Input and Output

```python
# Input features

X = data.iloc[:, :-1]

# Output labels

y = data['target']
```

# StandardScaler

```python
# Feature scaling

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

# Train-Test Split

```python
# Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

# KNN Model

```python
# Create KNN model

model = KNeighborsClassifier(n_neighbors=5)
```

# Train Model

```python
# Train model

model.fit(X_train, y_train)
```

# Predict Output

```python
# Predictions

y_pred = model.predict(X_test)

# Flower names

flower_names = iris.target_names

predicted_flower_classes = [
    flower_names[i]
    for i in y_pred
]

# Numeric Output

print("Predicted Numeric Classes:\n")

print(y_pred)

# Flower Names Output

print("\nPredicted Flower Classes:\n")

print(predicted_flower_classes)
```

# performance evaluation

```python
# Accuracy

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
# F1 Score

f1 = f1_score(y_test, y_pred, average='weighted')

print("F1 Score:", f1)
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Classification report

print(classification_report(y_test, y_pred))
```

# conclusion

The K-Nearest Neighbors (KNN) classification model was successfully implemented using the Iris dataset.

The dataset was preprocessed using StandardScaler and divided into 80% training data and 20% testing data.

The model achieved high performance with:
• Accuracy Score = 100%
• F1 Score = 100%

The Confusion Matrix and Classification Report confirmed that the model classified Setosa, Versicolor, and Virginica flowers correctly.
